## Zip files

```python
import zipfile
import os

def zip_files(file_paths, output_zip_name):
    """
    Compresses a list of files into a single zip archive.

    Args:
        file_paths (list): A list of paths to the files to be zipped.
        output_zip_name (str): The name of the output zip file (e.g., 'my_archive.zip').
    """
    with zipfile.ZipFile(output_zip_name, 'w', zipfile.ZIP_DEFLATED) as zipf:
        for file_path in file_paths:
            if os.path.exists(file_path):
                # arcname is the name of the file inside the zip archive.
                # By default, it's the full path, but you usually want just the filename.
                zipf.write(file_path, arcname=os.path.basename(file_path))
                print(f"Added {file_path} to {output_zip_name}")
            else:
                print(f"Warning: File not found - {file_path}")

if __name__ == "__main__":
    files_to_zip = ["ace2004_full_span-marker-bert-base-fewnerd-fine-super_20250424_194229.json",
                    "ace2004_full_span-marker-roberta-large-ontonotes5_20250424_194220.json",
                    "ace2004_full_span-marker-xlm-roberta-large-conllpp-doc-context_20250424_194152.json",
                    "ace_test_golden_annotations_with_candidates.json",
                    "aida_test_full_span-marker-bert-base-fewnerd-fine-super_20250424_194129.json",
                    "aida_test_full_span-marker-roberta-large-ontonotes5_20250424_194112.json",
                    "aida_test_full_span-marker-xlm-roberta-large-conllpp-doc-context_20250424_194021.json",
                    "aida_test_golden_annotations_with_candidates.json",
                    "aida_train_converted_span-marker-bert-base-fewnerd-fine-super_20250424_152936.json",
                    "aida_train_converted_span-marker-roberta-large-ontonotes5_20250424_152828.json",
                    "aida_train_converted_span-marker-xlm-roberta-large-conllpp-doc-context_20250424_163255.json",
                    "aida_train_golden_annotations_with_candidates.json",
                    "candidates_data_collector_20250424_104609.json",
                    ]
    output_zip = "cached_calculations.zip"

    zip_files(files_to_zip, output_zip)
    print(f"\nSuccessfully created {output_zip}")
```

## Unzip files

```python
import zipfile
import os

def unzip_files(zip_file_path, extract_dir):
    """
    Extracts all files from a zip archive to a specified directory.

    Args:
        zip_file_path (str): The path to the zip file to be unzipped.
        extract_dir (str): The directory where the files will be extracted.
    """
    try:
        with zipfile.ZipFile(zip_file_path, 'r') as zipf:
            # Create the extraction directory if it doesn't exist
            os.makedirs(extract_dir, exist_ok=True)
            zipf.extractall(extract_dir)
            print(f"Successfully extracted {zip_file_path} to {extract_dir}")
    except zipfile.BadZipFile:
        print(f"Error: {zip_file_path} is not a valid zip file.")
    except FileNotFoundError:
        print(f"Error: Zip file not found at {zip_file_path}")

if __name__ == "__main__":
    input_zip = "cached_calculations.zip"
    output_extract_dir = "cached_calculations"

    unzip_files(input_zip, output_extract_dir)

    # List the contents of the zip file
    print("\nContents of the zip file:")
    try:
        with zipfile.ZipFile(input_zip, 'r') as zipf:
            zipf.printdir()
    except FileNotFoundError:
        print(f"Error: Zip file not found at {input_zip}")
```
